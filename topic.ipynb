**Question:** In the 85 replicate EPMA analyses of reference material 7802 Apatite, does the reported `errpct` match the observed replicate scatter for each element?

**Data:** `data/EMPA_Apatite_Standards.csv`, from the USGS data release "Apatite and monazite geochemistry data at Mountain Pass, California, 2024" (Benson and Watts), https://doi.org/10.5066/P1LGUGBG, accessed 2026-11-03.

**Purpose:** load the data from a relative path and report its size, columns and units.

**Scientific choice:** the file has 86 data rows. 85 are replicate analyses of the reference material `7802 Apatite`, and one is the `Detection Limit` row, which is not a measurement and must not enter the replicate scatter later. The two are counted separately.

**Expected output:** 86 rows in the file, split into 85 replicate analyses and 1 detection-limit row; the 28 column names; and the unit of each kind of column.
